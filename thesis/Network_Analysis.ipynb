# PTSD Ketamine Study: Network Analysis

This notebook provides a foundation for analyzing the connectivity metrics extracted from the 37-participant cohort.

## Research Question
**Does baseline segregation between SN and DMN predict treatment response to ketamine/psychotherapy?**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# Load the dataset extracted by extract_metrics.py
df = pd.read_csv('thesis_dataset.csv')

# Preview the data
df.head()
```

## 1. Distribution of Segregation
Let's look at the System Segregation Index across different sessions (e.g., Pre vs Post).

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='session', y='system_segregation_index', data=df, hue='task')
plt.title('SN-DMN Segregation Index by Session and Task')
plt.ylabel('System Segregation Index (Chan et al., 2014)')
plt.show()
```

## 2. Statistical Comparison
Is there a significant change in segregation from Session 1 to Session 2 for the Resting State?

```python
# Example: Compare Rest task between Session 1 and Session 2
rest_df = df[df['task'] == 'rest']
ses1 = rest_df[rest_df['session'] == 'ses-01']['system_segregation_index']
ses2 = rest_df[rest_df['session'] == 'ses-02']['system_segregation_index']

if len(ses1) > 0 and len(ses2) > 0:
    t_stat, p_val = stats.ttest_rel(ses1, ses2)
    print(f"Paired t-test (Resting State): t={t_stat:.3f}, p={p_val:.4f}")
else:
    print("Not enough data for paired comparison yet.")
```
